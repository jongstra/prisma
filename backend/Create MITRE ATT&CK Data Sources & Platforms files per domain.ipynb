# Create MITRE ATT&CK Data Sources & Platforms files per domain

Prepare a tactics_and_techniques.json file before running this notebook.

```python
import json
```

```python
with open('tactics_and_techniques.json', 'r') as f:
  data = json.load(f)
```

```python
def flatten(lst):
    """
    Flattens a nested list, ignoring NoneType objects.

    Parameters:
    lst (list): A list that may contain nested lists and NoneType objects.

    Returns:
    list: A flattened list with NoneType objects removed.
    """
    result = []

    def _flatten(sublist):
        for item in sublist:
            if item is None:
                continue
            if isinstance(item, list):
                _flatten(item)
            else:
                result.append(item)

    _flatten(lst)
    return result
```

```python
platforms = {}
data_sources = {}
data_components = {}
for domain in data.keys():
    platforms[domain] = []
    data_sources[domain] = []
    data_components[domain] = []
    for tactic in data[domain]:
        for technique in tactic['techniques']:
            platforms[domain].append(technique.get('platforms'))
            data_sources[domain].append(technique.get('data_sources'))
            data_components[domain].append(technique.get('data_components'))
            if 'sub_technique' in technique.keys():
                for sub_technique in technique['sub_techniques']:
                    platforms[domain].append(sub_technique.get('platforms'))
                    data_sources[domain].append(sub_technique.get('data_sources'))
                    data_components[domain].append(sub_technique.get('data_components'))
    
    # Flatten lists.
    platforms[domain] = list(set(flatten(platforms[domain])))
    data_sources[domain] = list(set(flatten(data_sources[domain])))
    data_components[domain] = list(set(flatten(data_components[domain])))
```

```python
# Save results as json file.
with open('platforms.json', 'w') as f:
    json.dump(platforms, f)

with open('data_sources.json', 'w') as f:
    json.dump(data_sources, f)

with open('data_components.json', 'w') as f:
    json.dump(data_components, f)
```
